## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter data contains 30,000 pages. The average content age is 256.2 days, and 174 pages have not been updated for 180+ days. These numbers show that there are pages that may deserve review for a content refresh or improvement.



```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

url = "https://raw.githubusercontent.com/varshitha922/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print("Columns:", list(df.columns))
print("Average content age (days):", round(df["content_age_days"].mean(), 1))
print("Average days since last update:", round(df["days_since_last_update"].mean(), 1))
print("Pages not updated for 180+ days:", int((df["days_since_last_update"] >= 180).sum()))
```
